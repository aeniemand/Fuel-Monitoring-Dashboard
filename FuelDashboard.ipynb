```python
import geopandas as gpd
import leafmap
import rasterio
import numpy as np
```

```python
geojson_path = r"C:\Users\ashni\Documents\Spring2026\GIST215C\Leafmap\StudyAreaShapefile\StudyArea_50mbuffer.geojson"
slope_path = r"C:\Users\ashni\Documents\Spring2026\GIST215C\Leafmap\Slope_10m\slope_10m.tif"
aspect_path = r"C:\Users\ashni\Documents\Spring2026\GIST215C\Leafmap\Aspect_10m\aspect_10_v2.tif"
```

```python
gdf = gpd.read_file(geojson_path)

if gdf.crs is not None and gdf.crs.to_epsg() != 4326:
    gdf = gdf.to_crs(epsg=4326)

minx, miny, maxx, maxy = gdf.total_bounds
center_lat = (miny + maxy) / 2
center_lon = (minx + maxx) / 2

m = leafmap.Map(
    center=[center_lat, center_lon],
    zoom=11,
    height="700px"
)
```

```python
m.add_basemap("Esri.WorldImagery")


m.add_geojson(
    geojson_path,
    layer_name="Study Area"
)

m.add_layer_control()
```

```python
slope_class_path = r"C:\Users\ashni\Documents\Spring2026\GIST215C\Leafmap\Slope_10m\slope_classes_10m.tif"
aspect_class_path = r"C:\Users\ashni\Documents\Spring2026\GIST215C\Leafmap\Aspect_10m\aspect_classes_10m.tif"
```

```python
m.add_raster(
    slope_class_path,
    layer_name="Slope Classes",
    opacity=0.75,
    nodata=255
)

m.add_raster(
    aspect_class_path,
    layer_name="Aspect Classes",
    opacity=0.75,
    nodata=255
)
```

```python
m.add_legend(
    title="Slope Classes",
    legend_dict={
        "0–15° Gentle": "#ffffb2",
        "15–30° Moderate": "#fecc5c",
        "30–45° Steep": "#fd8d3c",
        "45–90° Very steep": "#e31a1c"
    }
)
```

```python
m.add_legend(
    title="Aspect Classes",
    legend_dict={
        "North": "#313695",
        "Northeast": "#4575b4",
        "East": "#74add1",
        "Southeast": "#abd9e9",
        "South": "#fee090",
        "Southwest": "#fdae61",
        "West": "#f46d43",
        "Northwest": "#a50026"
    }
)
```

```python
import ipywidgets as widgets
from IPython.display import display
```

```python
home_tab = widgets.HTML("""
<h2>Fuel Monitoring Dashboard</h2>

<p>
This dashboard is being developed to help identify potential fuel monitoring
priority areas using spatial layers related to terrain, fuels, vegetation
structure, fuel moisture and other environmental factors.
</p>

<p>
The current map includes terrain layers derived from a 10m DTM, including
slope and aspect. Additional layers will include LFMC, VPD from
the same years, canopy height, and fuel model/fuelscape data.
</p>

<p>
The goal is to use these layers together to explore where fuel structure and 
environmnetal conditions may contribute to higher intesity fire and help plan
priority areas for fuel reduction.
</p>
""")

methods_tab = widgets.HTML("""
<h2>Methods and Next Steps</h2>

<p>
Slope and aspect were derived from a clipped 10 m DTM and classified to make
terrain patterns easier to interpret.
</p>

<p>
The next phase of the analysis will add LFMC and VPD from the
same period. The VPD analysis will consider lagged relationships with LFMC to
better represent delayed fuel moisture response to atmospheric drying demand.
</p>

<p>
Because the input layers have different spatial resolutions, the combined
priority analysis will use a common analysis grid. Continuous variables can be
resampled or aggregated, while categorical layers such as aspect classes and
fuel model classes should use nearest neighbor so class values are preserved.
</p>

<p>
The final analysis will combine terrain, vegetation structure, fuel model,
LFMC and lagged VPD layers into a fuel monitoring priority index.
</p>
""")

tabs = widgets.Tab(children=[home_tab, m, methods_tab])
tabs.set_title(0, "Home")
tabs.set_title(1, "Map")
tabs.set_title(2, "Methods")
```
